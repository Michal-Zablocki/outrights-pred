```python
import pandas as pd
```

```python
TR = 3  # teams to be relegated
```

```python
df = pd.read_csv('data/20260220/ekstraklasa_relegation_odds_raw.csv')
```

```python
df.head()
```

```python
for bookie in ['Betclic', 'STS', 'efortuna']:
    spread = sum(1 / df[f'Odds - {bookie}'])
    if bookie == 'Betclic':
        df[f'Breakeven Prob - {bookie} - tax adj. [%]'] = df.apply(lambda x: round(100 / (x[f'Odds - {bookie}']), 1), axis=1)
    else:
        df[f'Breakeven Prob - {bookie} - tax adj. [%]'] = df.apply(lambda x: round(100 / ((x[f'Odds - {bookie}']) * 0.88), 1), axis=1)
    df[f'Implied Prob - {bookie} - raw [%]'] = df.apply(lambda x: round(100 / x[f'Odds - {bookie}'], 1), axis=1)
    df[f'Implied Prob - {bookie} - spread adj. [%]'] = df.apply(lambda x: round(100 / x[f'Odds - {bookie}'] * (TR / spread), 1), axis=1)
```
